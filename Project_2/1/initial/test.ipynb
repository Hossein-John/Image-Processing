```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('./sample/image_two.jpg')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
# convert BGR to HSV

hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
plt.imshow(cv2.cvtColor(hsv_img, cv2.COLOR_HSV2RGB))
plt.axis('off')
plt.show()
```

```python
lower_skin = np.array([0, 40, 60])
upper_skin = np.array([25, 180, 255])
mask = cv2.inRange(hsv_img, lower_skin, upper_skin)
# res = cv2.bitwise_and(img, img, mask=mask)

plt.imshow(cv2.cvtColor(mask, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
# --- IGNORE ---
```

```python
# morphology 
plt.imshow(cv2.cvtColor(mask, cv2.COLOR_BGR2RGB))
plt.title('Original Mask')
plt.axis('off')
plt.show()
kernel  = np.ones((3,3), np.uint8)


opening1 = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel, iterations = 2)
plt.imshow(cv2.cvtColor(opening1, cv2.COLOR_BGR2RGB))
plt.title('After Opening')
plt.axis('off')
plt.show()

closing1 = cv2.morphologyEx(opening1, cv2.MORPH_CLOSE, kernel, iterations = 2)
plt.imshow(cv2.cvtColor(closing1, cv2.COLOR_BGR2RGB))
plt.title('After Closing')
plt.axis('off')
plt.show()


opening2 = cv2.morphologyEx(closing1, cv2.MORPH_OPEN, kernel, iterations = 1)
plt.imshow(cv2.cvtColor(opening2, cv2.COLOR_BGR2RGB))
plt.title('After Opening')
plt.axis('off')
plt.show()



```

```python
final_img_RGB = cv2.cvtColor(opening2, cv2.COLOR_GRAY2RGB)
cv2.imwrite('image_two.jpg', final_img_RGB)
```
