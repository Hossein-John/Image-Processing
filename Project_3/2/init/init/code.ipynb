```python
contours, _ = cv2.findContours(
    closed_image,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

H, W = closed_image.shape[:2]
digit_boxes = []

for cnt in contours:
    x, y, w, h = cv2.boundingRect(cnt)
    area = w * h

    # فیلتر ساده‌ی ارقام
    if h > 0.4 * H and 0.01 * H * W < area < 0.1 * H * W:
        digit_boxes.append((x, y, w, h))

# 1️⃣ مرتب‌سازی از چپ به راست
digit_boxes = sorted(digit_boxes, key=lambda b: b[0])

# 2️⃣ حذف حرف وسط اگر وجود داشت
if len(digit_boxes) == 6:
    digit_boxes.pop(2)

# 3️⃣ نمایش فقط اعداد
vis = cropped_image.copy()
for x, y, w, h in digit_boxes:
    cv2.rectangle(vis, (x, y), (x+w, y+h), (0, 255, 0), 2)

plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
import numpy as np

def ssd(a, b):
    return np.sum((a.astype(np.int32) - b.astype(np.int32)) ** 2)
```

```python
def recognize_digit(digit_img, templates):
    best_digit = None
    best_score = float('inf')

    for d, tmpl in templates.items():
        score = ssd(digit_img, tmpl)
        if score < best_score:
            best_score = score
            best_digit = d

    return best_digit
```

```python
plate_digits = []

for (x, y, w, h) in digit_boxes:
    digit_img = closed_image[y:y+h, x:x+w]
    digit_img = cv2.resize(digit_img, (28, 28))
    pred = recognize_digit(digit_img, processed_templates)
    plate_digits.append(pred)

print("Plate digits:", plate_digits)
```
