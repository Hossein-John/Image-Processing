```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# ---- مرحله ۱: استرچینگ هیستوگرام روی هر کانال ----
def histogram_stretching(channel):
    min_val = np.min(channel)
    max_val = np.max(channel)
    stretched = (channel - min_val) * (255 / (max_val - min_val))
    return np.uint8(stretched)

r, g, b = cv2.split(image_rgb)
r_stretched = histogram_stretching(r)
g_stretched = histogram_stretching(g)
b_stretched = histogram_stretching(b)

image_stretched = cv2.merge([r_stretched, g_stretched, b_stretched])

# ---- مرحله ۲: اعمال گاما کرکشن ----
gamma = 0.7 # مقدار مناسب گاما
gamma_correction = lambda x: np.uint8((x / 255.0) ** gamma * 255)

image_gamma = cv2.merge([
    gamma_correction(r_stretched),
    gamma_correction(g_stretched),
    gamma_correction(b_stretched)
])

# ---- نمایش تصاویر ----
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(image_stretched)
plt.title("Histogram Stretched Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(image_gamma)
plt.title("Final Enhanced Image (Gamma)")
plt.axis("off")

plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import numpy as np

def histogram_stretching(image, output_min=0, output_max=255):
    # کمینه و بیشینه شدت روشنایی در تصویر اصلی
    input_min, input_max = np.min(image), np.max(image)
    # جلوگیری از تقسیم بر صفر در صورتی که تصویر کاملاً یکنواخت باشد
    if input_max == input_min:
        return np.full_like(image, output_min)

    # اعمال فرمول کشش هیستوگرام
    stretched = ((image - input_min) / (input_max - input_min)) * (output_max - output_min) + output_min
    # برش مقادیر برای مطمئن شدن از باقی ماندن در بازه
    stretched = np.clip(stretched, output_min, output_max)
    return stretched.astype(np.uint8)

# بارگذاری تصویر به صورت خاکستری
image = cv2.imread('37.jpeg', cv2.IMREAD_GRAYSCALE)

# اعمال کشش هیستوگرام
stretched_image = histogram_stretching(image)

# محاسبه هیستوگرام
original_hist = cv2.calcHist([image], [0], None, [256], [0, 256])
stretched_hist = cv2.calcHist([stretched_image], [0], None, [256], [0, 256])

# نمایش تصاویر و هیستوگرام‌ها
plt.figure(figsize=(15, 10))

# تصویر اصلی
plt.subplot(2, 2, 1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.axis('off')

# تصویر پردازش‌شده
plt.subplot(2, 2, 2)
plt.title("Histogram Stretched Image")
plt.imshow(stretched_image, cmap='gray')
plt.axis('off')

# هیستوگرام تصویر اصلی
plt.subplot(2, 2, 3)
plt.title("Histogram: Original Image")
plt.plot(original_hist, color='blue')
plt.xlim([0, 256])
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

# هیستوگرام تصویر پردازش‌شده
plt.subplot(2, 2, 4)
plt.title("Histogram: Stretched Image")
plt.plot(stretched_hist, color='orange')
plt.xlim([0, 256])
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np

def adaptive_histogram_equalization(img, clip_limit=2.0, tile_size=8):
    """
    Apply adaptive histogram equalization with customizable parameters
    """
    # Convert to uint8 if needed
    if img.dtype != np.uint8:
        img = (img * 255).astype(np.uint8)
    
    # Create CLAHE object with parameters
    clahe = cv2.createCLAHE(clipLimit=clip_limit, 
                           tileGridSize=(tile_size, tile_size))
    
    # Apply CLAHE
    enhanced = clahe.apply(img)
    return enhanced

def bi_histogram_equalization(img):
    """
    Apply bi-histogram equalization
    """
    # Calculate mean intensity
    mean = np.mean(img)
    
    # Separate the image into two parts
    low = img[img < mean]
    high = img[img >= mean]
    
    # Equalize each part separately
    if len(low) > 0:
        low = cv2.equalizeHist(low.astype(np.uint8))
    if len(high) > 0:
        high = cv2.equalizeHist(high.astype(np.uint8))
    
    # Combine the results
    result = np.zeros_like(img)
    result[img < mean] = low[0:len(result[img < mean])]
    result[img >= mean] = high[0:len(result[img >= mean])]
    
    return result

def gamma_equalization(img, gamma=1.1):
    """
    Apply gamma correction followed by histogram equalization
    """
    # Gamma correction
    gamma_corrected = np.power(img / 255.0, gamma) * 255.0
    gamma_corrected = gamma_corrected.astype(np.uint8)
    
    # Histogram equalization
    enhanced = cv2.equalizeHist(gamma_corrected)
    return enhanced

def weighted_equalization(img, weight=0.5):
    """
    Combine original and equalized image with weights
    """
    # Regular histogram equalization
    equalized = cv2.equalizeHist(img.astype(np.uint8))
    
    # Weighted combination
    result = cv2.addWeighted(img, 1-weight, equalized, weight, 0)
    return result.astype(np.uint8)

def enhance_medical_image(img, method='gamma', **params):
    """
    Enhance medical image using specified method
    
    Parameters:
    -----------
    img : numpy.ndarray
        Input image
    method : str
        One of 'adaptive', 'bi_histogram', 'gamma', 'weighted'
    params : dict
        Additional parameters for specific methods
    """
    if method == 'adaptive':
        return adaptive_histogram_equalization(
            img, 
            params.get('clip_limit', 2.0),
            params.get('tile_size', 8)
        )
    elif method == 'bi_histogram':
        return bi_histogram_equalization(img)
    elif method == 'gamma':
        return gamma_equalization(
            img,
            params.get('gamma', 1.2)
        )
    elif method == 'weighted':
        
        return weighted_equalization(
            img,
            params.get('weight', 0.5)
        )
    else:
        raise ValueError(f"Unknown method: {method}")
```

```python
import cv2
import numpy as np

def contrast_brightness_score(original_path, processed_path):
    # خواندن تصاویر
    original = cv2.imread(original_path)
    processed = cv2.imread(processed_path)

    # تبدیل تصاویر به فضای رنگی Lab
    original_lab = cv2.cvtColor(original, cv2.COLOR_BGR2Lab)
    processed_lab = cv2.cvtColor(processed, cv2.COLOR_BGR2Lab)

    # استخراج کانال L (روشنایی)
    L_original = original_lab[:, :, 0]
    L_processed = processed_lab[:, :, 0]

    # محاسبه حداقل و حداکثر مقدار روشنایی در تصاویر
    min_L_original, max_L_original = np.min(L_original), np.max(L_original)
    min_L_processed, max_L_processed = np.min(L_processed), np.max(L_processed)

    # محاسبه نمره روشنایی و کنتراست
    brightness_score = np.mean(L_processed) / np.mean(L_original)  # بررسی افزایش میانگین روشنایی
    contrast_score = (max_L_processed - min_L_processed) / (max_L_original - min_L_original)  # بررسی افزایش محدوده روشنایی

    return brightness_score, contrast_score

# نمونه اجرا
original_image_path = "9.png"
processed_image_path = "image_gamma.png"

brightness, contrast = contrast_brightness_score(original_image_path, processed_image_path)
print(f"Brightness Score: {brightness:.2f}")
print(f"Contrast Score: {contrast:.2f}")

# بررسی اینکه آیا بهبود یافته است؟
if brightness > 1.1 and contrast > 1.1:
    print("✅ تصویر بهبود یافته است!")
else:
    print("❌ تصویر بهبود نیافته یا به درستی پردازش نشده است.")
```
